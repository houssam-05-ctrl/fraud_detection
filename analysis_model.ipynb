```python
import os
print("PATH:", os.environ.get('PATH', 'Not set'))

# Try to find chmod the Python way
import shutil
chmod_path = shutil.which('chmod')
print(f"chmod location: {chmod_path}")

# Check if we can use chmod through Python
import subprocess
try:
    result = subprocess.run(['which', 'chmod'], capture_output=True, text=True)
    print(f"which chmod: {result.stdout}")
except:
    print("Could not run which command")
```
